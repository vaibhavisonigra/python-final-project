Data Visualizationn

```python
import matplotlib.pyplot as plt
import seaborn as sns

```

Top  Characters

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='character', palette='viridis')
plt.title('Most Popular  characters')
plt.xlabel('Characters')
plt.ylabel('Player Count')
plt.show()        
```

Rank Distribution 

```python
plt.Figure(figsize=(8, 5))
sns.countplot(data=df, x='rank', palette='magma')
plt.title('Player Distribution by Rank')
plt.xlabel('Rank')
plt.ylabel('Count')
plt.show()
```

Kills vs K/D RAtio (Scatter Plot):

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='kills', y='kd', hue='rank')
plt.title('Kills vs K/D Ratio by Rank')
plt.show()
```
